```python
unsortarr=np.array([3,4,2,1,5,9,7,6])
print(np.sort(unsortarr))
#sorting array 1d and 2d 
unsortarr_2d=np.array([[3,4],
                       [5,1],
                       [9,1]])
print(np.sort(unsortarr_2d,axis=0))#if you give axis is o it sort the whole colum and axis 1 sort the row vise data 
```

```python
arr_new=np.arange(20)
even_num=arr_new[arr_new % 2 ==0]#index value perfrom the opration aslo in numpy only not in python 
print(even_num)
```

```python
mask=arr_new > 10
#mask store the full exp. with its out put 
print(arr_new[mask])
```

```python
where_re=np.where(arr_new>10)
print(where_re)
print(arr_new[where_re])
```

```python
conditions=np.where(arr_new >10,arr_new,arr_new)
print(conditions)
```
