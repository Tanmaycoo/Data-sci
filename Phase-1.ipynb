```python
arr_new=np.arange(0,20)
reshaped = arr_new.reshape((4,5))#it re-shape the array in given size 
print("it re-shape the array in given size",reshaped)

flattened=arr_new.flatten()#it convert the array in orignal form 
print("it convert the array in orignal form ",flattened)

revel=arr_new.ravel() #it print the copy of the array 
print("it print the copy of the array ",revel)
```
