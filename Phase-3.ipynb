```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
arr=np.array([[1,2],[3,4]])
new_row=np.array([[3,4]])

new_array=np.vstack((arr,new_row))#add the values in the array in the row 
print(new_array)

new_col=np.array([[3,2],[4,5]])#add the values in array in the columns 
new_array_col=np.hstack((arr,new_col))
print(new_array_col)
```

```python
data=np.loadtxt("resturents.csv",delimiter=",")
data1 = data.astype(int)
```

```python
print(data1)
```

```python
sum=np.sum(data1[:,1:],axis=0)
print(sum)
```

```python
min_sale=np.min(data1[:,1:],axis=1)
print(min_sale)
```

```python
max_sale=np.max(data1[:,1:],axis=1)
print(max_sale)

cousum=np.cumsum(data1,axis=0)
```

```python
plt.figure(figsize=(6,3))
plt.plot(np.mean(cousum,axis=0))
plt.title(" Avg of Cumsum Sale per Year ")
plt.xlabel("Years")
plt.ylabel("Sale")
plt.grid(True)
```
